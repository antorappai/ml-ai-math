# Scaled Dot-Product Attention
Implement score, scale, softmax, and value mixing while annotating every shape.

```python
import torch
torch.manual_seed(42)
Q = torch.randn(2, 4, 8)
K = torch.randn(2, 4, 8)
V = torch.randn(2, 4, 8)
scores = Q @ K.transpose(-2, -1) / (Q.shape[-1] ** 0.5)
weights = torch.softmax(scores, dim=-1)
output = weights @ V
print('scores:', scores.shape, 'output:', output.shape)
print('row sum:', weights[0, 0].sum().item())
```

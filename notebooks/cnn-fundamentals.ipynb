# CNN Fundamentals
Track channels and spatial dimensions through convolution and pooling.

```python
import torch
torch.manual_seed(42)
X = torch.randn(8, 3, 32, 32)
model = torch.nn.Sequential(torch.nn.Conv2d(3, 16, 3, padding=1), torch.nn.ReLU(), torch.nn.MaxPool2d(2))
y = model(X)
print('input:', X.shape, 'output:', y.shape)
```

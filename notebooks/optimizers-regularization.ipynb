# Optimizers and Regularization
Compare SGD and Adam under a deterministic setup, then add weight decay.

```python
import torch
torch.manual_seed(42)
parameter = torch.nn.Parameter(torch.tensor([5.0]))
optimizer = torch.optim.Adam([parameter], lr=0.1, weight_decay=0.01)
for step in range(20):
    optimizer.zero_grad()
    loss = (parameter - 1).pow(2).mean()
    loss.backward()
    optimizer.step()
print('parameter:', round(parameter.item(), 4), 'loss:', round(loss.item(), 4))
```

# Backpropagation and Autograd
Compare the chain-rule result with PyTorch's automatic derivative.

```python
import torch
torch.manual_seed(42)
w = torch.tensor(3.0, requires_grad=True)
x = torch.tensor(2.0)
loss = (w * x - 5) ** 2
loss.backward()
manual = 2 * (w.detach() * x - 5) * x
print('autograd:', w.grad.item(), 'manual:', manual.item())
```
